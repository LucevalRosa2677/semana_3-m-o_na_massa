```python
# dicionário chamado veiculo com as chaves: modelo, ano e preco.
veiculo = {
    "modelo": "logan",
    "ano": 2023,
    "preco": 75000.00
}
print("Dicionário 'veiculo' inicial:")
print(veiculo)
```

```python
# Calcule o preço do veículo com um desconto de 10% e atualize o valor no dicionário.
desconto_percentual = 0.10
preco_com_desconto = veiculo["preco"] * (1 - desconto_percentual)
veiculo["preco"] = preco_com_desconto
print("\nDicionário 'veiculo' após aplicar 10% de desconto:")
print(veiculo)
```

```python
#  SET de CPFs com 3 números repetidos e imprima para provar que os repetidos sumiram.
cpfs_com_duplicatas = {"123.456.789-00", "987.654.321-11", "123.456.789-00", "111.222.333-44", "987.654.321-11","111.222.333-44", "555.456.789-07"}
print("\nSET de CPFs com duplicatas (antes de criar o set, para visualização):")
print(list(cpfs_com_duplicatas)) # Convert to list to show original items as entered if needed for context

# Ao criar um SET, as duplicatas são automaticamente removidas
cpfs_unicos = set(cpfs_com_duplicatas)
print("\nSET de CPFs após remoção de duplicatas (imprimindo o set):")
print(cpfs_unicos)
print("\nComprimento do SET original (com duplicatas):", len(cpfs_com_duplicatas))
print("Comprimento do SET após remoção de duplicatas (sem duplicatas):", len(cpfs_unicos))
```

```python
# Criando uma lista onde 3 CPFs estão repetidos (total de 6 entradas)
cpfs_com_repetidos = [
    "123.456.789-00",
    "987.654.321-11",
    "111.222.333-44",
    "123.456.789-00", # Repetido 1
    "123.456.789-00", # Repetido 2
    "987.654.321-11"  # Repetido 3
]

# Convertendo a lista para um SET para limpar as duplicatas
cpfs_unicos = set(cpfs_com_repetidos)

# Provando que os repetidos sumiram
print(f"Total na lista original: {len(cpfs_com_repetidos)}")
print(f"Total no SET final: {len(cpfs_unicos)}")
print("-" * 30)
print("CPFs que sobraram no SET:")
for cpf in cpfs_unicos:
    print(cpf)
```
